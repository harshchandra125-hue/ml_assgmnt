```python
import numpy as np
```

```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([[1, 2], [3, 4]])
c = np.array([[[1, 2]], [[3, 4]]])
```

```python
# shape functionm - returns dimension of the array
np.shape(a)
 
np.shape(np.array([[1, 2], [3, 4]]))

np.shape(np.zeros((3, 4)))
```

```python
# ncim function - returns dimension of the array (1, 2, 3)
a.ndim

b.ndim

c.ndim
```

```python
# size function - returns total number of elements
a.size

np.zeros((2, 3)).size

np.ones((2, 2, 3)).size
```

```python
# dtype function - returns data type
np.array([1, 2, 3]).dtype

np.array([1.5, 2.5, 3.5]).dtype

np.array(["A", "B", "C"]).dtype
```

```python
# reshape function - change thee shaoe of an array

a.reshape(3, 4)

a.reshape(2, 6)

a.reshape(-1, 2)
```

```python
# flatten function - comverts multidimension into 1 dimension array - creates a copy

a = np.array([[1, 2], [3, 4]])

a.flatten()

np.array([[1, 2, 3], [4, 5, 6]]).flatten()

np.arange(12).reshape(3, 4).flatten()
```

```python
# ravel function - converts multidimension into 1 dimension array - changes refernenced array

a = np.array([[1, 2], [3, 4]])

a.ravel()

np.arange(6).reshape(2, 3).ravel()

np.array([[[1, 2], [3, 4]]]).ravel()
```

```python
# transpose function - changes rows into columns and columns into rows

a = np.array([[1, 2, 3], [4, 5, 6]])

np.transpose(a)
a.T

np.array([[1, 2], [3, 4], [5, 6]]).transpose()
```

```python
# concatenate function - joins 2 arrays 

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

np.concatenate((a, b))

np.concatenate((a, b, [7, 8]))

np.concatenate((a.reshape(1, 3), b.reshape(1, 3)), axis=0)
```

```python
# sum function - calculates sum of array elements

a = np.array([10, 20, 30, 40])

np.sum(a)

np.sum(np.array([[1, 2], [3, 4]]))

np.sum(np.array([[1, 2], [3, 4]]), axis=0)
```

```python
# mean and median function 

a = np.array([10, 20, 30, 40])

np.mean(a)

np.mean(np.array([[1, 2], [3, 4]]))

np.mean(np.array([[1, 2], [3, 4]]), axis=0)

np.median([1, 3, 5, 7, 9])

np.median([10, 20, 30, 40])

np.median(np.array([[1, 2], [3, 4]]))
```

```python
# min and max function - returns minimum and maximum of array

a = np.array([10, 5, 30, 2, 20])

np.min(a)

np.min(np.array([[1, 8], [3, 4]]))

np.min(np.array([[1, 8], [3, 4]]), axis=0)

np.max(a)

np.max(np.array([[1, 8], [3, 4]]))

np.max(np.array([[1, 8], [3, 4]]), axis=0)
```

```python
# std and var function - returns standard deviation and variance

a = np.array([10, 20, 30, 40])

np.std(a)

np.std(np.array([[1, 2], [3, 4]]))

np.std(np.array([[1, 2], [3, 4]]), axis=0)

# Example 1
np.std(a)

np.std(np.array([[1, 2], [3, 4]]))

np.std(np.array([[1, 2], [3, 4]]), axis=0)
```

```python
# select function - chooses value from different choices based on conditions

a = np.array([10, 24, 30, 40, 52, 66, 89, 99, 11, 33])

cond = [(a>90), ((a<90)&(a>70)), ((a<70)&(a>50)), ((a<50)&(a>30))]
choice = ["A", "B", "C", "D"]
print(np.select(cond, choice, default="F"))
```
